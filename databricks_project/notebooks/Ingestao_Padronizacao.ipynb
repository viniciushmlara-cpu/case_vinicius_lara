```python
%pip install rapidfuzz
from rapidfuzz import process, fuzz

from pyspark.sql import functions as F
from pyspark.sql.types import StringType
import unidecode
import re

def remove_accents(text):
    if text is None:
        return None
    return unidecode.unidecode(text)

remove_accents_udf = F.udf(remove_accents, StringType())

df = spark.table("case_mevo.prescricaomedicamento")

df = df.withColumn(
    "cidadepaciente_clean",
    remove_accents_udf(F.col("cidadepaciente"))
)

df = df.withColumn(
    "cidadepaciente_final",
    F.initcap(F.col("cidadepaciente_clean"))
)

# Load municipios table and create a dictionary of cities per UF
municipios_df = spark.table("workspace.case_mevo.municipios")
municipios_dict = (
    municipios_df
    .groupBy("UF")
    .agg(F.collect_list("Municipio").alias("cities"))
    .collect()
)
uf_cities = {row["UF"]: row["cities"] for row in municipios_dict}

# Fuzzy matching function
def fuzzy_city(cidade, uf):
    if cidade is None or uf is None:
        return None
    # Special case: treat "SP" as "São Paulo" for UF "SP"
    if cidade.strip().upper() == "SP" and uf == "SP":
        return "Sao Paulo"
    # Handle invalid city names: only digits, single character, or known invalids
    if re.fullmatch(r"\d+", cidade) or (len(cidade) < 3 and cidade.strip().upper() != "SP") or cidade.lower() in ["abc", "x", "-"]:
        return None
    cities = uf_cities.get(uf, [])
    if not cities:
        return None
    return process.extractOne(
        cidade,
        cities,
        scorer=fuzz.WRatio
    )[0]

fuzzy_city_udf = F.udf(fuzzy_city, StringType())

# List of valid Brazilian state abbreviations
valid_states = [
    "RO", "RR", "GO", "PA", "AP", "TO", "MT", "MA", "AC", "MS", "AM", "PI", "RN", "PB", "PE",
    "AL", "MG", "ES", "RJ", "SP", "PR", "SC", "RS", "CE", "SE", "BA", "DF"
]

# Update estadopaciente only if not in valid_states or if null (do not use estado_medico)
df = df.withColumn(
    "estadopaciente_corrigido",
    F.when(
        (F.col("estadopaciente").isNull()) | (~F.col("estadopaciente").isin(valid_states)),
        None
    ).otherwise(F.col("estadopaciente"))
)

# Fuzzy match cidadepaciente_final, do not fallback to cidade_medico_final
df = df.withColumn(
    "cidadepadronizada",
    F.when(
        F.col("cidadepaciente_final").isNotNull() & (F.length(F.col("cidadepaciente_final")) > 0),
        fuzzy_city_udf(F.col("cidadepaciente_final"), F.col("estadopaciente_corrigido"))
    ).otherwise(None)
)

# Join with cidadesclassificadas to get classe_populacao
cidadesclassificadas_df = spark.table("workspace.case_mevo.cidadesclassificadas")
df = df.join(
    cidadesclassificadas_df,
    (F.col("cidadepadronizada") == F.col("municipio")) & (F.col("estadopaciente_corrigido") == F.col("uf")),
    "left"
)

df = df.withColumn(
    "sexopaciente_padronizado",
    F.when(F.col("sexopaciente").isin(["F", "Feminino"]), "F")
     .when(F.col("sexopaciente").isin(["M", "Masculino"]), "M")
     .otherwise(None)
)

# Adiciona o campo min_dataprescricao por idpaciente
from pyspark.sql.window import Window
w = Window.partitionBy("idprescricao")
df = df.withColumn("min_dataprescricao", F.min("dataprescricao").over(w))

# Select and rename columns, incluindo min_dataprescricao
df_silver = df.select(
    "idprescricao","idmedicamento","dataprescricao","idmedico","idpaciente","idconvenio","nascimentopaciente",
    F.col("sexopaciente_padronizado").alias("sexopaciente"),
    F.col("estadopaciente_corrigido").alias("estadopaciente"),
    F.col("cidadepadronizada").alias("cidadepaciente"),
    "visualizadapaciente","permitirimpressao","itemvendido","datavenda","canalvenda",
    F.col("classe_populacao").alias("grupocidadepaciente"),
    "min_dataprescricao"
)

display(df_silver)

# Save as Delta table in the silver layer
df_silver.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.case_mevo_silver.prescricaomedicamento")
```

```python
from pyspark.sql import functions as F
from pyspark.sql.types import StructType, StructField, StringType
import unidecode
import re
from rapidfuzz import process, fuzz

def remove_accents(text):
    if text is None:
        return None
    return unidecode.unidecode(text)
remove_accents_udf = F.udf(remove_accents, StringType())

municipios_df = spark.table("workspace.case_mevo.municipios")

municipios_dict = (
    municipios_df
    .groupBy("UF")
    .agg(F.collect_list("Municipio").alias("cities"))
    .collect()
)
uf_cities = {row["UF"]: row["cities"] for row in municipios_dict}

def fuzzy_city_uf(cidade):
    if cidade is None or len(cidade.strip()) == 0:
        return (None, None)
    cidade_clean = unidecode.unidecode(cidade)
    cidade_clean = cidade_clean.strip().title()
    best_score = 0
    best_city = None
    best_uf = None
    for uf, cities in uf_cities.items():
        match = process.extractOne(cidade_clean, cities, scorer=fuzz.WRatio)
        if match and match[1] > best_score:
            best_score = match[1]
            best_city = match[0]
            best_uf = uf
    if best_score >= 80:
        return (best_city, best_uf)
    return (None, None)

fuzzy_city_uf_udf = F.udf(
    fuzzy_city_uf,
    StructType([
        StructField("cidade_ajustada", StringType(), True),
        StructField("uf_ajustada", StringType(), True)
    ])
)

medicos_df = spark.table("workspace.case_mevo.medicos")

medicos_df = medicos_df.withColumn(
    "genero_corrigido",
    F.when(F.col("genero").isin(["F", "M", "Feminino", "Masculino"]),
           F.when(F.col("genero").isin(["F", "Feminino"]), "F")
            .when(F.col("genero").isin(["M", "Masculino"]), "M")
            .otherwise(None)
    ).otherwise(None)
)

medicos_df = medicos_df.withColumn(
    "cidade_medico_clean",
    remove_accents_udf(F.col("cidade"))
)
medicos_df = medicos_df.withColumn(
    "cidade_medico_final",
    F.initcap(F.col("cidade_medico_clean"))
)

medicos_df = medicos_df.withColumn(
    "fuzzy_struct",
    fuzzy_city_uf_udf(F.col("cidade_medico_final"))
)
medicos_df = medicos_df.withColumn("cidade_ajustada", F.col("fuzzy_struct.cidade_ajustada"))
medicos_df = medicos_df.withColumn("uf_ajustada", F.col("fuzzy_struct.uf_ajustada"))

conselho_map = {
    "CRM": "CRM",
    "CRO": "CRO",
    "CRM-ES": "CRM",
    "CRM - RJ": "CRM",
    "CROSP": "CRO",
    "CRM - SP": "CRM",
    "CRP": "CRP - Psicologia",
    "COREN": "COREN - Enfermagem",
    "CRN": "CRN - Nutricionista",
    "116158": "CRM",
    "CREFITO": "CREFITO - Fisio e Terapia Ocupacional",
    "CRM-GO": "CRM",
    "CRO-GO": "CRO",
    "CRM - CE": "CRM",
    "CRM - ES": "CRM",
    "CRORJ": "CRO",
    "CRM - PE": "CRM",
    "CRF": "CRF - Farmacia",
    "OUT": "",
    "RMS": "RMS - Mais Medicos",
    "CRFA": "CRFA - Fonoaudiologo",
    "ABQ": "ABQ - Quiropraxia",
    "CRO-ES": "CRO",
    "ADM": "",
    "---": "",
    "CREFONO": "CREFONO - Fono",
    "CRTR": "CRTR - Tec Radiologia",
    "CRN-3": "CRN - Nutricionista",
    "OUTROS": "",
    "CRBM": "CRBM - Biomedicina",
    "CRESS": "CRESS - Servico Social",
    "CRMRES": "CRM",
    "CRM - PA": "CRM",
    "CRN3": "CRN - Nutricionista",
    "CRMV": "CRMV",
    "CRA": "CRA - Adm Saude",
    "CREMESP": "CRM",
    "CFP": "CFP - Psicologia",
    "CAUSP": ""
}
conselho_map_expr = F.create_map([F.lit(x) for x in sum(conselho_map.items(), ())])
medicos_df = medicos_df.withColumn(
    "conselhopadronizado",
    conselho_map_expr.getItem(F.col("conselhoprofissional"))
)

# Adiciona coluna grupoespecialidade cruzando especialidade com especialidade_raw
grupoespecialidade_df = spark.table("workspace.case_mevo.grupoespecialidade")
medicos_df = medicos_df.join(
    grupoespecialidade_df,
    medicos_df["especialidade"] == grupoespecialidade_df["especialidade_raw"],
    "left"
)

medicos_final = medicos_df.select(
    "idmedico", "conselhoprofissional",
    "conselhopadronizado",
    "ufconselho",
    F.col("genero_corrigido").alias("genero"),
    "idade", 
    F.col("cidade_ajustada").alias("cidade"),
    F.col("uf_ajustada").alias("estado"),
    "especialidade",
    "grupoespecialidade"
)

display(medicos_final)

medicos_final.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.case_mevo_silver.medicos")
```

```python
from pyspark.sql.functions import when, lit

medicamentos_df = spark.table("workspace.case_mevo.medicamentos")

medicamentos_df = medicamentos_df.withColumn(
    "portaria344",
    when(
        medicamentos_df["nome"].isin(["Cloridrato de Fluoxetina", "Arcoxia"]),
        "LISTA - C1"
    ).otherwise(medicamentos_df["portaria344"])
)

medicamentos_df = medicamentos_df.withColumn(
    "controleespecial",
    when(medicamentos_df["nome"].isin(["Codein", "Donaren Retard", "Arcoxia"]), "true")
    .when(medicamentos_df["nome"] == "Paracetamol", "false")
    .otherwise(medicamentos_df["controleespecial"])
)

preco_dict = {
    8156: 29.09,
    9366: 22.99,
    20254: 57.49,
    42541: 14.05,
    14714: 8.09,
    11341: 133.59,
    41452: 26.71,
    1124: 39.52,
    5783: 28.56,
    13501: 5.58,
    27752: 44.49,
    7387: 127.99,
    2950: 91.98,
    3180: 78.49,
    1492: 22.56,
    8903: 40.89,
    11191: 43.55,
    42402: 15.59,
    9832: 82.99,
    3783: 64.99,
    30811: 24.16,
    38479: 39.06
}

preco_expr = when(lit(False), None)
for idmed, preco in preco_dict.items():
    preco_expr = preco_expr.when(medicamentos_df["idmedicamento"] == idmed, lit(preco))
preco_expr = preco_expr.otherwise(None)

medicamentos_df = medicamentos_df.withColumn("preco", preco_expr)

display(medicamentos_df)

medicamentos_df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.case_mevo_silver.medicamentos")
```

```python
from pyspark.sql.functions import (
    countDistinct,
    count,
    weekofyear,
    col,
    when,
    year,
    max as spark_max,
    hour,
    to_date,
    datediff
)

# Carregar tabelas
df = spark.table("workspace.case_mevo_silver.prescricaomedicamento")
medicos_df = spark.table("workspace.case_mevo_silver.medicos").select("idmedico", "especialidade")
medicamentos_df = spark.table("workspace.case_mevo_silver.medicamentos").select(
    "idmedicamento", "controleespecial", "antimicrobiano", "portaria344", "mip", col("nome").alias("nomemedicamento"), col("preco").alias("preco_medicamento"), "impressaoduasvias"
)

# Join com médicos e medicamentos
joined_df = df.join(
    medicos_df,
    on="idmedico",
    how="left"
).join(
    medicamentos_df,
    on="idmedicamento",
    how="left"
)

# Adicionar coluna de idade
joined_df = joined_df.withColumn(
    "idade",
    when((year(col("nascimentopaciente")) >= 1901) & (year(col("nascimentopaciente")) <= 1927), "1 - Greatest")
    .when((year(col("nascimentopaciente")) >= 1928) & (year(col("nascimentopaciente")) <= 1944), "2 - Silenciosa")
    .when((year(col("nascimentopaciente")) >= 1945) & (year(col("nascimentopaciente")) <= 1964), "3 - Baby boomers")
    .when((year(col("nascimentopaciente")) >= 1965) & (year(col("nascimentopaciente")) <= 1981), "4 - Geração X")
    .when((year(col("nascimentopaciente")) >= 1982) & (year(col("nascimentopaciente")) <= 1994), "5 - Millennials")
    .when((year(col("nascimentopaciente")) >= 1995) & (year(col("nascimentopaciente")) <= 2010), "6 - Geração Z")
    .when(year(col("nascimentopaciente")) > 2010, "7 - Alfa")
)

# Adicionar coluna plano_saude: 1 se idconvenio não é nulo, senão 0
joined_df = joined_df.withColumn(
    "plano_saude",
    when(col("idconvenio").isNotNull(), 1).otherwise(0)
)

# Adicionar coluna regiao_brasil baseada no estado do paciente
joined_df = joined_df.withColumn(
    "regiao_brasil",
    when(col("estadopaciente").isin("AC", "AP", "AM", "PA", "RO", "RR", "TO"), "Norte")
    .when(col("estadopaciente").isin("AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"), "Nordeste")
    .when(col("estadopaciente").isin("DF", "GO", "MT", "MS"), "Centro-Oeste")
    .when(col("estadopaciente").isin("ES", "MG", "RJ", "SP"), "Sudeste")
    .when(col("estadopaciente").isin("PR", "RS", "SC"), "Sul")
)

# Marcação de prescrições especiais
from pyspark.sql import Window

window_prescricao = Window.partitionBy("idprescricao")

joined_df = joined_df.withColumn(
    "itemcontroleespecial",
    spark_max(when(col("controleespecial") == "true", 1).otherwise(0)).over(window_prescricao)
).withColumn(
    "itemantimicrobiano",
    spark_max(when(col("antimicrobiano") == "true", 1).otherwise(0)).over(window_prescricao)
)

# Adicionar coluna numero_semana_ano
joined_df = joined_df.withColumn("numero_semana_ano", weekofyear("min_dataprescricao"))

# Adicionar coluna periodo_prescricao baseada em dataprescricao
joined_df = joined_df.withColumn(
    "periodo_prescricao",
    when((hour(col("dataprescricao")) >= 5) & (hour(col("dataprescricao")) < 12), "manha")
    .when((hour(col("dataprescricao")) >= 12) & (hour(col("dataprescricao")) < 18), "tarde")
    .when((hour(col("dataprescricao")) >= 18) & (hour(col("dataprescricao")) < 23), "noite")
    .otherwise("madrugada")
)

# Adicionar coluna com a diferença entre data da prescricao e data da venda
joined_df = joined_df.withColumn(
    "dias_entre_prescricao_e_venda",
    datediff(
        to_date(col("datavenda")),
        to_date(col("min_dataprescricao"))
    )
)

# Selecionar apenas as colunas desejadas
cols_to_drop = [
    "nascimentopaciente", "estadopaciente", "cidadepaciente", "itemlistaA2", "itemlistaC1"
]
final_cols = [c for c in joined_df.columns if c not in cols_to_drop]

final_df = joined_df.select(final_cols)

display(final_df)

final_df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("case_mevo_gold.cubo_prescricao")
```
